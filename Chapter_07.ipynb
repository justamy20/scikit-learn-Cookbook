## 1. Konsep Dasar SVM: Maximum Margin Classifier
Tujuan utama SVM adalah mencari sebuah **Hyperplane** (garis pemisah) yang memiliki **Margin** terbesar antara dua kelas. Titik-titik data yang paling dekat dengan garis pemisah ini disebut sebagai **Support Vectors**.

Secara matematis, SVM mencoba memaksimalkan jarak:
$$\frac{2}{||w||}$$
Subjek terhadap: $y_i(w^T x_i + b) \geq 1$

```python
import numpy as np
from sklearn import svm
from sklearn.datasets import make_blobs
from sklearn.model_selection import train_test_split

# Membuat dataset yang bisa dipisahkan secara linear
X, y = make_blobs(n_samples=50, centers=2, random_state=6)

# Inisialisasi SVM dengan Kernel Linear
# Parameter C: Mengontrol keseimbangan antara margin lebar dan klasifikasi yang benar
clf = svm.SVC(kernel='linear', C=1000)
clf.fit(X, y)

print(f"Jumlah Support Vectors: {clf.support_vectors_.shape[0]}")
print(f"Indeks Support Vectors: {clf.support_})")
```

---
## 2. Kernel Trick: Menangani Data Non-Linear
Jika data tidak bisa dipisahkan dengan garis lurus, SVM menggunakan **Trik Kernel**. Teknik ini secara matematis memetakan data ke dimensi yang lebih tinggi di mana data tersebut menjadi dapat dipisahkan secara linear.

Beberapa jenis Kernel yang populer:
1. **Polynomial Kernel:** $K(x, y) = (x^T y + r)^d$
2. **RBF (Radial Basis Function) Kernel:** $K(x, y) = \exp(-\gamma ||x - y||^2)$

```python
from sklearn.datasets import make_circles

# Membuat data berbentuk lingkaran (tidak bisa dipisahkan garis lurus)
X_c, y_c = make_circles(n_samples=100, factor=0.1, noise=0.1, random_state=42)

# Mencoba SVM Linear (pasti akurasinya rendah)
clf_linear = svm.SVC(kernel='linear').fit(X_c, y_c)

# Menggunakan SVM RBF (mampu menangani pola lingkaran)
clf_rbf = svm.SVC(kernel='rbf', gamma='auto').fit(X_c, y_c)

print(f"Akurasi SVM Linear pada data lingkaran: {clf_linear.score(X_c, y_c) * 100:.2f}%")
print(f"Akurasi SVM RBF pada data lingkaran   : {clf_rbf.score(X_c, y_c) * 100:.2f}%")
```

---
## 3. Parameter Penting: C dan Gamma
Untuk mendapatkan performa terbaik dengan SVM, kita harus mengatur dua hyperparameter utama:

* **Parameter C:** Mengontrol penalti terhadap kesalahan klasifikasi.
    * *C Besar:* Margin kecil, mencoba mengklasifikasikan semua titik training dengan benar (risiko overfitting).
    * *C Kecil:* Margin lebih lebar, lebih toleran terhadap kesalahan (lebih stabil/general).
* **Parameter Gamma (khusus RBF):** Menentukan seberapa jauh pengaruh satu titik data tunggal.
    * *Gamma Besar:* Pengaruh titik dekat saja (garis keputusan sangat meliuk-liuk).
    * *Gamma Kecil:* Pengaruh titik sampai jauh (garis keputusan lebih halus).

```python
# Demonstrasi pengaruh C
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# C Kecil vs C Besar
clf_low_c = svm.SVC(kernel='linear', C=0.01).fit(X_train, y_train)
clf_high_c = svm.SVC(kernel='linear', C=100.0).fit(X_train, y_train)

print(f"Akurasi C=0.01 (Test): {clf_low_c.score(X_test, y_test):.4f}")
print(f"Akurasi C=100 (Test) : {clf_high_c.score(X_test, y_test):.4f}")
```
