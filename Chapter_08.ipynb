## 1. Decision Tree (Pohon Keputusan)
Decision Tree bekerja dengan membagi data berdasarkan fitur yang paling informatif menggunakan kriteria seperti **Gini Impurity** atau **Entropy**. Pohon ini sangat mudah diinterpretasikan karena menyerupai logika manusia "Jika-Maka".



[Image of Decision Tree structure for classification]


<div class="alert alert-warning">
<b>Kekurangan Utama:</b> Decision Tree tunggal cenderung mengalami <b>Overfitting</b> (terlalu menghafal data training). Oleh karena itu, kita jarang menggunakannya sendirian untuk data yang kompleks.
</div>

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.datasets import load_wine
from sklearn.model_selection import train_test_split

# Load dataset Wine (Klasifikasi jenis anggur)
wine = load_wine()
X, y = wine.data, wine.target
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Inisialisasi Decision Tree
# max_depth=3 digunakan untuk mencegah pohon tumbuh terlalu kompleks (overfitting)
tree_clf = DecisionTreeClassifier(max_depth=3, random_state=42)
tree_clf.fit(X_train, y_train)

print(f"Akurasi Decision Tree (Train): {tree_clf.score(X_train, y_train):.4f}")
print(f"Akurasi Decision Tree (Test) : {tree_clf.score(X_test, y_test):.4f}")
```

---
## 2. Random Forest (Bootstrap Aggregating)
Random Forest adalah kumpulan dari banyak Decision Tree yang dibangun secara acak. Teknik ini disebut **Bagging**. Setiap pohon memberikan suara (vote), dan suara terbanyak menjadi prediksi akhir.



Kelebihan Random Forest:
1. Jauh lebih stabil dibanding Decision Tree tunggal.
2. Mengurangi risiko overfitting secara signifikan.
3. Dapat menghitung **Feature Importance** (fitur mana yang paling berpengaruh).

```python
from sklearn.ensemble import RandomForestClassifier

# Inisialisasi Random Forest dengan 100 pohon (n_estimators)
rf_clf = RandomForestClassifier(n_estimators=100, random_state=42)
rf_clf.fit(X_train, y_train)

print(f"Akurasi Random Forest (Test): {rf_clf.score(X_test, y_test):.4f}")

# Menampilkan 3 fitur paling penting menurut Random Forest
import pandas as pd
import numpy as np

importances = rf_clf.feature_importances_
indices = np.argsort(importances)[::-1]
print("\n3 Fitur Paling Berpengaruh:")
for i in range(3):
    print(f"{i+1}. {wine.feature_names[indices[i]]} ({importances[indices[i]]:.4f})")
```

---
## 3. Boosting (AdaBoost dan Gradient Boosting)
Berbeda dengan Bagging yang membangun pohon secara paralel, **Boosting** membangun pohon secara berurutan. Setiap pohon baru mencoba memperbaiki kesalahan yang dibuat oleh pohon sebelumnya.

* **AdaBoost**: Memberikan bobot lebih pada data yang salah diklasifikasikan sebelumnya.
* **Gradient Boosting**: Menggunakan fungsi loss gradient untuk meminimalkan kesalahan secara bertahap.

```python
from sklearn.ensemble import AdaBoostClassifier, GradientBoostingClassifier

# 1. AdaBoost
ada_clf = AdaBoostClassifier(n_estimators=50, random_state=42)
ada_clf.fit(X_train, y_train)

# 2. Gradient Boosting
gb_clf = GradientBoostingClassifier(n_estimators=50, learning_rate=0.1, random_state=42)
gb_clf.fit(X_train, y_train)

print(f"Akurasi AdaBoost: {ada_clf.score(X_test, y_test):.4f}")
print(f"Akurasi Gradient Boosting: {gb_clf.score(X_test, y_test):.4f}")
```
